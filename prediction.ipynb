```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np 
```

```python
## Load all the Trained Model Scaler,encoded pickle and model files

## Model file
model=load_model('model.h5')

# Loading Label Encoder pickle file
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)

with open('onehot_encoder_geo.pkl','rb') as file:
    oh_geo=pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data={
    'CreditScore':600, 
    'Geography':'France',
      'Gender':'Male', 
      'Age':40,
        'Tenure':3, 
        'Balance':60000,
       'NumOfProducts':2,
         'HasCrCard':1,
           'IsActiveMember':1, 
           'EstimatedSalary':50000,
}
```

```python
## one-hot Encode 'Geography'
geo_encoded=oh_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=oh_geo.get_feature_names_out(['Geography']))
```

```python
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
```

```python
## Encoding the gender value
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
```

```python
input_df
```

```python
# Concatenation
input_df=pd.concat([input_df.drop('Geography',axis=1),geo_encoded_df],axis=1)
```

```python
input_df
```

```python
## Scaling the data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## Predict churn
prediction=model.predict(input_scaled)
```

```python
prediction_proba=prediction[0][0]
prediction_proba
```

```python
if prediction_proba>0.5:
    print("The Customer is likely to churn")
else:
    print("The Customer is not likey to Churn")
```
